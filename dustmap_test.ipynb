# Dustmaps3d Unit Test

Testing the Wang+2025 3D dust map with known sightlines to verify units and behavior.

```python
import numpy as np

try:
    from dustmaps3d import dustmaps3d
    print("dustmaps3d loaded")
except ImportError:
    !pip install dustmaps3d
    from dustmaps3d import dustmaps3d
```

## Test 1: Known Sightlines

| Sightline | l | b | Expected E(B-V) |
|-----------|---|---|------------------|
| Galactic Center | 0 | 0 | HIGH (>1 mag at few kpc) |
| Galactic Anticenter | 180 | 0 | Moderate |
| North Galactic Pole | 0 | 90 | LOW (<0.05 mag) |
| South Galactic Pole | 0 | -90 | LOW (<0.05 mag) |
| Cygnus (high extinction) | 80 | 0 | HIGH |

```python
# Test sightlines at various distances
# Assuming distance is in kpc as documented

test_cases = [
    # (name, l, b, distances_kpc)
    ("Galactic Center (l=0, b=0)", 0.0, 0.0, [0.5, 1.0, 2.0, 5.0]),
    ("Galactic Anticenter (l=180, b=0)", 180.0, 0.0, [0.5, 1.0, 2.0, 5.0]),
    ("North Galactic Pole (b=+90)", 0.0, 90.0, [0.5, 1.0, 2.0, 5.0]),
    ("South Galactic Pole (b=-90)", 0.0, -90.0, [0.5, 1.0, 2.0, 5.0]),
    ("Cygnus region (l=80, b=0)", 80.0, 0.0, [0.5, 1.0, 2.0, 5.0]),
    ("Mid-latitude (l=0, b=30)", 0.0, 30.0, [0.5, 1.0, 2.0, 5.0]),
]

print("Testing dustmaps3d with distance in KPC (as documented)")
print("="*70)

for name, l, b, distances in test_cases:
    print(f"\n{name}")
    print("-" * 50)
    
    for d in distances:
        l_arr = np.array([l])
        b_arr = np.array([b])
        d_arr = np.array([d])
        
        EBV, dust, sigma, max_d = dustmaps3d(l_arr, b_arr, d_arr)
        
        # Extract values
        ebv_val = EBV.values[0] if hasattr(EBV, 'values') else EBV[0]
        maxd_val = max_d.values[0] if hasattr(max_d, 'values') else max_d[0]
        
        A_V = 3.1 * ebv_val
        
        print(f"  d={d:4.1f} kpc: E(B-V)={ebv_val:6.3f} mag, A_V={A_V:6.3f} mag, max_d={maxd_val:.1f} kpc")
```

## Test 2: What if distance is actually in PARSECS?

```python
print("Testing dustmaps3d with distance in PARSECS (if docs are wrong)")
print("="*70)

# Same sightlines but with distances as if they were in pc
test_cases_pc = [
    ("Galactic Center (l=0, b=0)", 0.0, 0.0, [500, 1000, 2000, 5000]),  # pc
    ("North Galactic Pole (b=+90)", 0.0, 90.0, [500, 1000, 2000, 5000]),
]

for name, l, b, distances in test_cases_pc:
    print(f"\n{name}")
    print("-" * 50)
    
    for d in distances:
        l_arr = np.array([l])
        b_arr = np.array([b])
        d_arr = np.array([d])  # Treating as pc
        
        EBV, dust, sigma, max_d = dustmaps3d(l_arr, b_arr, d_arr)
        
        ebv_val = EBV.values[0] if hasattr(EBV, 'values') else EBV[0]
        maxd_val = max_d.values[0] if hasattr(max_d, 'values') else max_d[0]
        
        A_V = 3.1 * ebv_val
        
        print(f"  d={d:5.0f} pc: E(B-V)={ebv_val:6.3f} mag, A_V={A_V:6.3f} mag, max_d={maxd_val:.1f}")
```

## Test 4: Re-query a few stars from your catalog

Manually query dustmaps3d for a few stars and compare.

```python
from astropy.coordinates import SkyCoord
import astropy.units as u

if enriched_files:
      catalog_file = enriched_files[0]
      with fits.open(catalog_file) as hdul:
          data = hdul[1].data

      # Convert ra/dec to Galactic
      coords = SkyCoord(ra=data['ra']*u.deg, dec=data['dec']*u.deg, frame='icrs')
      gal = coords.galactic
      gal_l = gal.l.deg
      gal_b = gal.b.deg

      # Pick 5 random stars
      n_test = min(5, len(data))
      idx = np.random.choice(len(data), n_test, replace=False)

      print("Re-querying dustmaps3d for sample stars:")
      print("="*80)

      for i in idx:
          l = gal_l[i]
          b = gal_b[i]
          plx = data['parallax'][i]

          if plx <= 0 or not np.isfinite(plx):
              continue

          dist_kpc = 1.0 / plx

          # Query dustmaps3d
          EBV, _, sigma, max_d = dustmaps3d([l], [b], [dist_kpc])
          ebv_val = EBV.values[0] if hasattr(EBV, 'values') else EBV[0]
          maxd_val = max_d.values[0] if hasattr(max_d, 'values') else max_d[0]
          A_V = 3.1 * ebv_val

          print(f"\nStar {i}: ra={data['ra'][i]:.3f}, dec={data['dec'][i]:.3f}")
          print(f"  -> l={l:.1f}, b={b:.1f}, plx={plx:.3f} mas, d={dist_kpc:.2f} kpc")
          print(f"  -> E(B-V)={ebv_val:.4f}, A_V={A_V:.3f} mag, max_d={maxd_val:.1f} kpc")
```
